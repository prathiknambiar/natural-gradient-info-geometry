```python
import numpy as np

np.random.seed(42)

true_mu = 3.0
data = np.random.normal(true_mu, 1.0, 100)

def log_likelihood(mu, data):
    return np.sum(-0.5 * (data - mu) ** 2)

def vanilla_gradient(mu, data):
    return np.mean(data - mu)

fisher = 1.0

def natural_gradient(mu, data):
    return vanilla_gradient(mu, data) / fisher

mu_vanilla = 0.0
mu_natural = 0.0

learning_rate = 0.1
iterations = 50

vanilla_history = []
natural_history = []

for i in range(iterations):
    mu_vanilla += learning_rate * vanilla_gradient(mu_vanilla, data)
    mu_natural += learning_rate * natural_gradient(mu_natural, data)

    vanilla_history.append(mu_vanilla)
    natural_history.append(mu_natural)

print("True Mean:", true_mu)
print("Sample Mean:", np.mean(data))
print("Vanilla Estimate:", mu_vanilla)
print("Natural Estimate:", mu_natural)

print("\nLog-Likelihood")
print("Vanilla:", log_likelihood(mu_vanilla, data))
print("Natural:", log_likelihood(mu_natural, data))

print("\nIteration\tVanilla\t\tNatural")
for i in range(iterations):
    print(f"{i+1}\t\t{vanilla_history[i]:.6f}\t{natural_history[i]:.6f}")
```
